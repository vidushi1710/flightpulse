```python
flights_clean = flights[[
    'YEAR', 'MONTH', 'DAY', 'DAY_OF_WEEK',
    'AIRLINE', 'ORIGIN_AIRPORT', 'DESTINATION_AIRPORT',
    'SCHEDULED_DEPARTURE', 'SCHEDULED_ARRIVAL',
    'DISTANCE', 'ARRIVAL_DELAY', 'DIVERTED', 'CANCELLED'
]].copy()

print(flights_clean.shape)
```

```python
flights_clean = flights_clean[
    (flights_clean['CANCELLED'] == 0) & (flights_clean['DIVERTED'] == 0)
].copy()

print(flights_clean.shape)
print(flights_clean['ARRIVAL_DELAY'].isnull().sum())
```

```python
print(flights_clean.isnull().sum())
```

```python
flights_clean['IS_DELAYED'] = (flights_clean['ARRIVAL_DELAY'] >= 15).astype(int)
print(flights_clean['IS_DELAYED'].value_counts())
print(flights_clean['IS_DELAYED'].value_counts(normalize=True))
```

```python
flights_clean['ORIGIN_AIRPORT'].apply(type).value_counts()
```

```python
flights_clean['FLIGHT_DATE'] = pd.to_datetime(
    flights_clean[['YEAR', 'MONTH', 'DAY']]
)
print(flights_clean['FLIGHT_DATE'].min(), "to", flights_clean['FLIGHT_DATE'].max())
```

```python
print(flights_clean.duplicated().sum())
```

```python
flights_clean['ORIGIN_AIRPORT'] = flights_clean['ORIGIN_AIRPORT'].astype(str)
flights_clean['DESTINATION_AIRPORT'] = flights_clean['DESTINATION_AIRPORT'].astype(str)

flights_clean = flights_clean.drop_duplicates()
print(flights_clean.shape)
```

```python
flights_clean.to_csv("../data/processed/flights_cleaned.csv", index=False)
print("Saved:", flights_clean.shape)
```

```python
Q1 = flights_clean['ARRIVAL_DELAY'].quantile(0.25)
Q3 = flights_clean['ARRIVAL_DELAY'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1, "Q3:", Q3, "IQR:", IQR)
print("Lower bound:", lower_bound, "Upper bound:", upper_bound)

outliers = flights_clean[
    (flights_clean['ARRIVAL_DELAY'] < lower_bound) | 
    (flights_clean['ARRIVAL_DELAY'] > upper_bound)
]
print("Number of outliers:", outliers.shape[0])
print("Percent of data:", round(outliers.shape[0] / flights_clean.shape[0] * 100, 2), "%")
```

```python
print("Minimum delay:", flights_clean['ARRIVAL_DELAY'].min())
print("Maximum delay:", flights_clean['ARRIVAL_DELAY'].max())
print()
print(flights_clean['ARRIVAL_DELAY'].describe())
```

```python
print("Overall delay statistics:")
print(flights_clean['ARRIVAL_DELAY'].agg(['mean', 'var', 'std']))
print()
print("Per airline (sorted worst to best):")
airline_stats = flights_clean.groupby('AIRLINE')['ARRIVAL_DELAY'].agg(['mean', 'std', 'count'])
print(airline_stats.sort_values('mean', ascending=False))
```

```python
# Extract departure hour from SCHEDULED_DEPARTURE (format is HHMM as an integer, e.g. 1430 = 2:30 PM)
flights_clean['DEP_HOUR'] = (flights_clean['SCHEDULED_DEPARTURE'] // 100).astype(int)

correlation_cols = ['DISTANCE', 'DAY_OF_WEEK', 'DEP_HOUR', 'ARRIVAL_DELAY']
corr_matrix = flights_clean[correlation_cols].corr()
print(corr_matrix['ARRIVAL_DELAY'])
```

```python
prob_by_airline = flights_clean.groupby('AIRLINE')['IS_DELAYED'].mean().sort_values(ascending=False)
print("P(Delayed | Airline):")
print(prob_by_airline)
```

```python
import pandas as pd

# Load raw data
flights = pd.read_csv("../data/raw/flights.csv")

# Select needed columns
flights_clean = flights[[
    'YEAR', 'MONTH', 'DAY', 'DAY_OF_WEEK',
    'AIRLINE', 'ORIGIN_AIRPORT', 'DESTINATION_AIRPORT',
    'SCHEDULED_DEPARTURE', 'SCHEDULED_ARRIVAL',
    'DISTANCE', 'ARRIVAL_DELAY', 'DIVERTED', 'CANCELLED'
]].copy()

# Remove cancelled/diverted flights
flights_clean = flights_clean[
    (flights_clean['CANCELLED'] == 0) & (flights_clean['DIVERTED'] == 0)
].copy()

# Fix airport code types
flights_clean['ORIGIN_AIRPORT'] = flights_clean['ORIGIN_AIRPORT'].astype(str)
flights_clean['DESTINATION_AIRPORT'] = flights_clean['DESTINATION_AIRPORT'].astype(str)

# Drop duplicates
flights_clean = flights_clean.drop_duplicates()

# Build target variable
flights_clean['IS_DELAYED'] = (flights_clean['ARRIVAL_DELAY'] >= 15).astype(int)

# Build departure hour
flights_clean['DEP_HOUR'] = (flights_clean['SCHEDULED_DEPARTURE'] // 100).astype(int)

print(flights_clean.shape)
```
